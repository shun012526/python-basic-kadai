```python
class Human:
  def __init__(self, name: str, age: int):
    self.name = name
    self.age = age

  def check_adult(self) -> str:
    if self.age >= 20:
      print(f"{self.name}は大人です。")
    else:
      print(f"{self.name}は大人ではありません。")


users = {}
users[0] = Human(name="Bob", age=19)
users[1] = Human(name="Tom", age=20)
users[2] = Human(name="Ken", age=21)
users[3] = Human(name="Akari", age=22)

for i in users:
  user = users[i]
  user.check_adult()
```
